Q1. Student Result Manager | Beginner
Create a Student class with a public name, protected _marks, and private __grade attributes. Initialize the
student through a constructor. Implement set_marks(marks), which accepts only values from 0 to 100, along with
get_marks() and get_grade(). Assign grade A for 80 to 100, B for 60 to 79, C for 40 to 59, and F below 40.
Demonstrate both a valid and an invalid marks update. Extension: Attempt direct access to __grade and explain
the result.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self._marks = 0
        self.__grade = "F"
        self.set_marks(marks)

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self._marks = marks

            if marks >= 80:
                self.__grade = "A"
            elif marks >= 60:
                self.__grade = "B"
            elif marks >= 40:
                self.__grade = "C"
            else:
                self.__grade = "F"
        else:
            print("Invalid marks! Enter 0 to 100.")

    def get_marks(self):
        return self._marks

    def get_grade(self):
        return self.__grade


student1 = Student("Ahmed", 85)
student2 = Student("Faizan", 55)

print("Student:", student1.name)
print("Marks:", student1.get_marks())
print("Grade:", student1.get_grade())

print("\nStudent:", student2.name)
print("Marks:", student2.get_marks())
print("Grade:", student2.get_grade())

student1.set_marks(95)
print("\nUpdated Marks:", student1.get_marks())

student1.set_marks(120)

try:
    print(student1.__grade)
except AttributeError:
    print("Private attribute cannot be accessed directly.")
```

Q2. Secure Bank Account | Intermediate
Create a BankAccount class with a public account holder name, private __balance, and private __pin. Implement
deposit(amount), withdraw(amount, pin), and get_balance(pin). Reject zero or negative deposits, withdrawals
exceeding the available balance, and incorrect PINs. Demonstrate a successful deposit, a successful withdrawal,
and two rejected operations. Extension: Explain why double underscores do not provide true security.

```python
class BankAccount:

    def __init__(self, name, balance, pin):
        self.name = name
        self.__balance = balance
        self.__pin = pin

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful:", amount)
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print("Incorrect PIN!")
        elif amount <= 0:
            print("Invalid withdrawal amount!")
        elif amount > self.__balance:
            print("Insufficient Balance!")
        else:
            self.__balance -= amount
            print("Withdrawal Successful:", amount)

    def get_balance(self, pin):
        if pin == self.__pin:
            return self.__balance
        else:
            print("Incorrect PIN!")
            return None


account1 = BankAccount("Ahmed", 10000, 1234)
account2 = BankAccount("Faizan", 15000, 5678)

account1.deposit(5000)
account1.withdraw(2000, 1234)

account1.deposit(-500)
account1.withdraw(1000, 1111)

print("\nAccount Holder:", account1.name)
print("Available Balance:", account1.get_balance(1234))

print("\nAccount Holder:", account2.name)
print("Available Balance:", account2.get_balance(5678))
```

Q3. Employee Roles | Beginner
Create a parent class Employee that stores a name and employee ID and provides display_details(). Derive
Developer (programming language) and Manager (team size). Use super() to initialize inherited attributes.
Override display_details() in each child class to include both shared and role-specific information. Create and test
one object of each type.


```python
class Employee:

    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id

    def display_details(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)


class Developer(Employee):

    def __init__(self, name, employee_id, language):
        super().__init__(name, employee_id)
        self.language = language

    def display_details(self):
        print("\nDeveloper Details")
        super().display_details()
        print("Programming Language:", self.language)


class Manager(Employee):

    def __init__(self, name, employee_id, team_size):
        super().__init__(name, employee_id)
        self.team_size = team_size

    def display_details(self):
        print("\nManager Details")
        super().display_details()
        print("Team Size:", self.team_size)


dev = Developer("Ahmed", 101, "Python")
manager = Manager("Faizan", 102, 10)

dev.display_details()
manager.display_details()
```

Q4. Vehicle Rental System | Intermediate
Create a base class Vehicle with brand, model, and daily rental rate. Implement calculate_rent(days). Derive Car,
which adds a fixed insurance fee of Rs. 1,000 per rental, and Bike, which applies a 10% discount to the base
rental amount. Override calculate_rent() and reuse the parent implementation with super(). Reject rental
durations of zero or fewer days. Test both vehicle types for a three-day rental.


```python
class Vehicle:

    def __init__(self, brand, model, daily_rate):
        self.brand = brand
        self.model = model
        self.daily_rate = daily_rate

    def calculate_rent(self, days):
        if days <= 0:
            print("Invalid rental duration!")
            return None

        return self.daily_rate * days


class Car(Vehicle):

    def calculate_rent(self, days):
        rent = super().calculate_rent(days)

        if rent is not None:
            return rent + 1000

        return None


class Bike(Vehicle):

    def calculate_rent(self, days):
        rent = super().calculate_rent(days)

        if rent is not None:
            return rent * 0.90

        return None


car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CD 70", 1500)

print("Ahmed's Car Rental:", car.calculate_rent(3))
print("Faizan's Bike Rental:", bike.calculate_rent(3))

car.calculate_rent(0)
```

Q5. Notification Dispatcher | Beginner
Create EmailNotification, SMSNotification, and PushNotification classes. Each must implement send(message)
with output specific to its channel. Write a single function notify_all(notifications, message) that iterates through
different notification objects and invokes send(). Add a WhatsAppNotification class without changing the
dispatcher. Explain where polymorphism occurs.

```python
class EmailNotification:

    def send(self, message):
        print("Email Notification:", message)


class SMSNotification:

    def send(self, message):
        print("SMS Notification:", message)


class PushNotification:

    def send(self, message):
        print("Push Notification:", message)


class WhatsAppNotification:

    def send(self, message):
        print("WhatsApp Notification:", message)


def notify_all(notifications, message):
    for notification in notifications:
        notification.send(message)


email = EmailNotification()
sms = SMSNotification()
push = PushNotification()
whatsapp = WhatsAppNotification()

notifications = [email, sms, push, whatsapp]

notify_all(notifications, "Ahmed and Faizan, your order is confirmed!")
```

Q6. Checkout Discount Calculator | Intermediate
Create RegularCustomer, PremiumCustomer, and VIPCustomer classes, each implementing
final_price(amount). Apply discounts of 5%, 10%, and 20%, respectively. Write print_receipt(customer, amount)
so that it works with any customer type. For a bill of Rs. 10,000, show the final amount for each type. Reject
negative bill amounts. Extension: Add GoldCustomer with a 15% discount without modifying print_receipt().

```python
class RegularCustomer:

    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount * 0.95


class PremiumCustomer:

    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount * 0.90


class VIPCustomer:

    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount * 0.80


class GoldCustomer:

    def final_price(self, amount):
        if amount < 0:
            print("Invalid bill amount!")
            return None

        return amount * 0.85


def print_receipt(customer, amount):
    price = customer.final_price(amount)

    if price is not None:
        print("Final Price: Rs.", price)


regular = RegularCustomer()
premium = PremiumCustomer()
vip = VIPCustomer()
gold = GoldCustomer()

print("Ahmed - Regular Customer")
print_receipt(regular, 10000)

print("\nFaizan - Premium Customer")
print_receipt(premium, 10000)

print("\nAhmed - VIP Customer")
print_receipt(vip, 10000)

print("\nFaizan - Gold Customer")
print_receipt(gold, 10000)

print("\nInvalid Bill Test")
print_receipt(regular, -1000)
```

Q7. Shape Calculator | Beginner
Use ABC and @abstractmethod to define an abstract Shape class with area() and perimeter(). Implement both
methods in Circle and Rectangle. Accept dimensions through constructors and use pi = 3.14159. Test a circle
with radius 5 and a rectangle with width 4 and height 6. Attempt to instantiate Shape directly and explain the
outcome.


```python
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14159 * self.radius


class Rectangle(Shape):

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Ahmed's Circle")
print("Area:", circle.area())
print("Perimeter:", circle.perimeter())

print("\nFaizan's Rectangle")
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())

try:
    shape = Shape()
except TypeError:
    print("\nShape cannot be instantiated directly.")
```

Q8. Payment Gateway | Intermediate
Define an abstract PaymentMethod class with abstract methods pay(amount) and refund(amount). Implement
CardPayment, WalletPayment, and BankTransferPayment. Each method must display the payment method,
transaction type, and amount. Validate that transaction amounts are positive. Write process_transaction(method,
amount) to call pay() for any supported payment method. Extension: Add a new payment method without
changing process_transaction().


```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class CardPayment(PaymentMethod):

    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Card Payment - Payment:", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Card Payment - Refund:", amount)


class WalletPayment(PaymentMethod):

    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Wallet Payment - Payment:", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Wallet Payment - Refund:", amount)


class BankTransferPayment(PaymentMethod):

    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Bank Transfer - Payment:", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Bank Transfer - Refund:", amount)


class CashPayment(PaymentMethod):

    def pay(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        else:
            print("Cash Payment - Payment:", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Invalid refund amount!")
        else:
            print("Cash Payment - Refund:", amount)


def process_transaction(method, amount):
    method.pay(amount)


card = CardPayment()
wallet = WalletPayment()
bank = BankTransferPayment()
cash = CashPayment()

print("Ahmed's Transaction")
process_transaction(card, 5000)

print("\nFaizan's Transaction")
process_transaction(wallet, 3000)

print("\nBank Transfer")
process_transaction(bank, 7000)

print("\nCash Payment")
process_transaction(cash, 2000)

print("\nRefund Example")
card.refund(1000)

print("\nInvalid Transaction")
process_transaction(wallet, -500)
```
