Q1. Accessing List Elements
numbers = [10, 20, 30, 40, 50, 60, 70]
Extract the first, middle, and last elements without modifying the original list.

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

first = numbers[0]
middle = numbers[len(numbers) // 2]
last = numbers[-1]

print("First:", first)
print("Middle:", middle)
print("Last:", last)
```

Q2. List Slicing
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
Create three new lists: first five elements, last four elements, and every second element in reverse order.


```python
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

first_five = numbers[:5]

last_four = numbers[-4:]

reverse_second = numbers[::-2]

print("First Five:", first_five)
print("Last Four:", last_four)
print("Every Second Reverse:", reverse_second)
```

Q3. List Modification
fruits = ["apple", "banana", "cherry", "mango"]
Replace banana with orange, add grapes at index 2, and remove cherry.

```python
fruits = ["apple", "banana", "cherry", "mango"]

fruits[1] = "orange"

fruits.insert(2, "grapes")

fruits.remove("cherry")

print(fruits)
```

Q4. Reverse Without reverse()
numbers = [12, 45, 23, 67, 89, 34]
Reverse using slicing only; preserve the original list.

```python
numbers = [12, 45, 23, 67, 89, 34]

reversed_list = numbers[::-1]

print("Original:", numbers)
print("Reversed:", reversed_list)
```

Q5. Second Largest Unique Number
numbers = [45, 12, 89, 34, 89, 67, 45, 23]
Find the second-largest unique number. Expected: 67.

```python
numbers = [45, 12, 89, 34, 89, 67, 45, 23]

unique_numbers = list(set(numbers))

unique_numbers.sort()

second_largest = unique_numbers[-2]

print("Second Largest:", second_largest)
```

Q6. Rotate a List
numbers = [10, 20, 30, 40, 50, 60]
k = 2
Rotate the list right by k using slicing. Expected: [50, 60, 10, 20, 30, 40].

```python
numbers = [10, 20, 30, 40, 50, 60]

k = 2

k = k % len(numbers)

rotated = numbers[-k:] + numbers[:-k]

print(rotated)
```

Q9. Extract Data from a Matrix
matrix = [
 [10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]
]
Extract main diagonal, secondary diagonal, and middle column; calculate the sum of each.

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diagonal = [
    matrix[0][0],
    matrix[1][1],
    matrix[2][2]
]

secondary_diagonal = [
    matrix[0][2],
    matrix[1][1],
    matrix[2][0]
]

middle_column = [
    matrix[0][1],
    matrix[1][1],
    matrix[2][1]
]

print("Main Diagonal:", main_diagonal)
print("Main Sum:", sum(main_diagonal))

print("Secondary Diagonal:", secondary_diagonal)
print("Secondary Sum:", sum(secondary_diagonal))

print("Middle Column:", middle_column)
print("Middle Sum:", sum(middle_column))
```

Q10. List Transformation Challenge
data = [15, 5, 25, 10, 35, 20, 30]
Build a list of the three largest unique values ascending, then two smallest unique values descending. Expected: [25,
30, 35, 10, 5].

```python
data = [15, 5, 25, 10, 35, 20, 30]

unique_data = list(set(data))

unique_data.sort()

largest_three = unique_data[-3:]

smallest_two = unique_data[:2][::-1]

result = largest_three + smallest_two

print(result)
```
